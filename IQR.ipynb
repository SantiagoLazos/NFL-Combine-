```python
def valoresreales(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    limINF = Q1 - 2 * IQR #Multiplico por 2 para quedarme con más de 8781 registros, si lo hacia con el normal de 1.5 se reducia en más de mil
    limSup = Q3 + 2 * IQR

    antes = len(df)
    df_filtrado = df[(df[col] >= limINF) & (df[col] <= limSup)]
    despues = len(df_filtrado)

    print(f"{col}: antes = {antes}, después = {despues}, eliminados = {antes - despues}")

    return df_filtrado
```

```python
cols_seguras = [
    "40 Yard",
    "Bench Press",
    "Vert Leap (in)",
    "Broad Jump (in)",
    "Weight (lbs)",
    "Height (in)"
]
for c in cols_seguras:
    df = valoresreales(df, c)
```
